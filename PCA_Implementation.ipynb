```python
#PCA of a data matrix

def my_pca(data , num_components, scree = False):




  # num_components = 2
  # data = X




  X = data - data.mean(axis = 0)





  cov = X.T@X

  evs = lg.eig(cov)

  print(evs)


  l = list(enumerate(evs[0]))
  ev_keys = sorted(l , key = lambda x:x[1],reverse=True )

  reduced_data = np.zeros((X.shape[0] , num_components))



  for i in range(num_components):

    j = ev_keys[i][0]

    pc_i = evs[1][:,j]

    reduced_data[: , i] = X@pc_i







  print(reduced_data.shape)


  if scree == True:

    plt.plot(sorted(evs[0], reverse = True))
    plt.show()


  return reduced_data
```

```python
reduced_data = my_pca(data = X , num_components = 2,scree = True)
```
